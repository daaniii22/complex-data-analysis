# **Pipeline de Construcción del Dataset Multimodal (2008–2026)**

Este notebook implementa el pipeline de extracción, corrección temporal y alineación multimodal entre:
- **Variables Biofísicas y Económicas:** NOAA ($CO_2$, $CH_4$, $N_2O$, ONI), NASA GISTEMP y futuros del petróleo crudo Brent.
- **Cobertura Mediática:** GDELT Summary API (volumen relativo semanal).
- **Narrativa Pública:** Revisiones históricas del artículo *Climate change* de Wikipedia.

> **Nota metodológica y marco teórico:**  
> La formulación de hipótesis estadísticas ($H_1, H_2$), la justificación de la granularidad temporal semanal y el diseño para análisis causal (Granger / VAR) se encuentran detallados en [`Readme.md`](./Readme.md).

```python
import datetime
import io
import time
import pandas as pd
import requests
import yfinance as yf
import tqdm
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from pathlib import Path
```

```python
HEADERS = {"User-Agent": "ResearchDataProject/1.0 (academic_nlp_study; contact: student@univ.edu)"}
session = requests.Session()
session.headers.update(HEADERS)

# Eje temporal que usaremos (Semanas: 2008 - Presente)
START_DATE = "2008-01-01"
END_DATE = datetime.date.today().isoformat()

# Evita el sesgo de festivos de Yahoo Finance creando un índice equiespaciado continuo
timeline = pd.date_range(start=START_DATE, end=END_DATE, freq="W-SUN", name="date")
df_base = pd.DataFrame(index=timeline).reset_index()
```

```python
# Funciones de extracción (Datos Físicos, Económicos y Mediáticos)
def get_noaa_co2():
    url = "https://gml.noaa.gov/webdata/ccgg/trends/co2/co2_weekly_mlo.csv"
    res = session.get(url, timeout=15)
    lines = [l for l in res.text.splitlines() if not l.strip().startswith("#")]
    df = pd.read_csv(io.StringIO("\n".join(lines)))
    df["date"] = pd.to_datetime(df[["year", "month", "day"]])
    return df[df["average"] > 0][["date", "average"]].rename(columns={"average": "co2_ppm"})


def get_brent_oil():
    # Descarga semanal del precio de cierre ajustado.
    data = yf.download(
        "BZ=F",
        start=START_DATE,
        end=END_DATE,
        interval="1wk",
        progress=False,
        auto_adjust=False,
    )
    if data.empty:
        return pd.DataFrame(columns=["date", "brent_price"])

    # yfinance puede devolver MultiIndex incluso al descargar un solo ticker.
    close = data["Close"]
    if isinstance(close, pd.DataFrame):
        close = close.iloc[:, 0]
    df = close.rename("brent_price").rename_axis("date").reset_index()
    df["date"] = pd.to_datetime(df["date"]).dt.tz_localize(None)
    return df.dropna(subset=["brent_price"])


def get_nasa_gistemp():
    # Anomalía mensual de temperatura global.
    # Evitamos el Look-Ahead Bias: el dato mensual de mes M se indexa al día 1 del mes M+1
    url = "https://data.giss.nasa.gov/gistemp/tabledata_v4/GLB.Ts+dSST.csv"
    res = session.get(url, timeout=15)
    df_raw = pd.read_csv(io.StringIO(res.text), skiprows=1, na_values="***")
    month_cols = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
    df = df_raw.melt(id_vars="Year", value_vars=month_cols, var_name="month", value_name="temp_anomaly_c")
    df["date"] = pd.to_datetime(df["Year"].astype(str) + df["month"], format="%Y%b") + pd.offsets.MonthBegin(1)
    df["temp_anomaly_c"] = pd.to_numeric(df["temp_anomaly_c"], errors="coerce")
    return df.dropna(subset=["temp_anomaly_c"])[["date", "temp_anomaly_c"]]


def get_noaa_ch4():
    url = "https://gml.noaa.gov/webdata/ccgg/trends/ch4/ch4_mm_gl.txt"
    res = session.get(url, timeout=15)
    lines = [l for l in res.text.splitlines() if not l.strip().startswith("#")]
    df = pd.read_csv(io.StringIO("\n".join(lines)), sep=r"\s+", usecols=[0, 1, 3], names=["year", "month", "ch4_ppb"])
    df["date"] = pd.to_datetime(df[["year", "month"]].assign(day=1)) + pd.offsets.MonthBegin(1)
    return df[["date", "ch4_ppb"]]


def get_noaa_n2o():
    url = "https://gml.noaa.gov/aftp/data/hats/n2o/combined/GML_global_N2O.txt"
    res = session.get(url, timeout=15)
    lines = [l for l in res.text.splitlines() if not l.strip().startswith("#")]
    df = pd.read_csv(io.StringIO("\n".join(lines)), sep=r"\s+", na_values="nan")
    df["date"] = pd.to_datetime(df[["GML_N2O_YYYY", "GML_N2O_MM"]].rename(columns={"GML_N2O_YYYY": "year", "GML_N2O_MM": "month"}).\
        assign(day=1)) + pd.offsets.MonthBegin(1)
    return df[["date", "GML_Global_N2O"]].rename(columns={"GML_Global_N2O": "n2o_ppb"}).dropna()


def get_noaa_oni():
    url = "https://www.cpc.ncep.noaa.gov/data/indices/oni.ascii.txt"
    res = session.get(url, timeout=15)
    df = pd.read_csv(io.StringIO(res.text), sep=r"\s+")
    season_to_month = {"DJF": 1, "JFM": 2, "FMA": 3, "MAM": 4, "AMJ": 5, "MJJ": 6, "JJA": 7, "JAS": 8, "ASO": 9, "SON": 10, "OND": 11, "NDJ": 12}
    df["month"] = df["SEAS"].map(season_to_month)
    df["date"] = pd.to_datetime(df[["YR", "month"]].rename(columns={"YR": "year"}).assign(day=1)) + pd.offsets.MonthBegin(1)
    return df[["date", "ANOM"]].rename(columns={"ANOM": "oni_anomaly"})


def get_gdelt_summary():
    # Cobertura mediática global normalizada.
    url = "https://api.gdeltproject.org/api/v2/summary/summary?d=web&q=climate%20change&format=json&mode=timelinevol&timelinesmooth=0"
    try:
        res = session.get(url, timeout=20).json()
        df = pd.DataFrame(res.get("timeline", [{}])[0].get("data", []))
        df["date"] = pd.to_datetime(df["date"])
        df["media_volume_norm"] = pd.to_numeric(df["value"], errors="coerce")
        return df[["date", "media_volume_norm"]]
    except Exception:
        return pd.DataFrame(columns=["date", "media_volume_norm"])
```

```python
# Alineamos todas las series temporales con la base semanal
data_loaders = [
    ("CO2", get_noaa_co2),
    ("Brent", get_brent_oil),
    ("GISTEMP", get_nasa_gistemp),
    ("CH4", get_noaa_ch4),
    ("N2O", get_noaa_n2o),
    ("ONI", get_noaa_oni),
    ("GDELT", get_gdelt_summary),
]

df_merged = df_base.copy()
for name, loader in tqdm.tqdm(data_loaders):
    print(f"Descargando y acoplando {name}...")
    df_src = loader().sort_values("date")
    if not df_src.empty:
        df_merged = pd.merge_asof(df_merged, df_src, on="date", direction="backward")
```

```python
# Extraemos las revisiones históricas de Wikipedia para el artículo "Climate change"
WIKI_ENDPOINT = "https://en.wikipedia.org/w/api.php"
ARTICLE_TITLE = "Climate_change"

def get_wiki_snapshot(title, target_date, max_retries=5):
    params = {
        "action": "query",
        "prop": "revisions",
        "titles": title,
        "rvlimit": 1,
        "rvstart": target_date.strftime("%Y-%m-%dT23:59:59Z"),
        "rvdir": "older",
        "rvsection": 0,
        "rvprop": "content|timestamp|ids|size",
        "format": "json",
    }
    for attempt in range(max_retries):
        try:
            r = session.get(WIKI_ENDPOINT, params=params, timeout=15)
            if r.status_code == 429:
                time.sleep(int(r.headers.get("Retry-After", 5)))
                continue
            r.raise_for_status()
            page = next(iter(r.json().get("query", {}).get("pages", {}).values()))
            rev = page.get("revisions", [{}])[0]
            return {
                "rev_id": rev.get("revid"),
                "rev_timestamp": rev.get("timestamp"),
                "rev_size": rev.get("size"),
                "lead_text": rev.get("*", ""),
            }
        except requests.RequestException:
            time.sleep(2 * (attempt + 1))
    return {"rev_id": None, "rev_timestamp": None, "rev_size": None, "lead_text": ""}

print(f"Extrayendo revisiones de Wikipedia para {len(df_merged)} semanas.")
wiki_records = []
for idx, row in tqdm.tqdm(df_merged.iterrows(), total=len(df_merged)):
    wiki_records.append(get_wiki_snapshot(ARTICLE_TITLE, row["date"]))

    if (idx + 1) % 50 == 0 or (idx + 1) == len(df_merged):
        print(f"Progreso: [{idx + 1}/{len(df_merged)} semanas]")

    time.sleep(0.5)

dataset_final = pd.concat([df_merged.reset_index(drop=True), pd.DataFrame(wiki_records)], axis=1)
dataset_final
```

```python
# Serializamos el dataset bruto consolidado a un archivo Parquet.

raw_dir = Path("data/raw")
raw_dir.mkdir(parents=True, exist_ok=True)
raw_output_path = raw_dir / "dataset_clima_multimodal.parquet"
dataset_final.to_parquet(raw_output_path, index=False)
print("Registros guardados:", len(dataset_final))
print("Archivo:", raw_output_path)
```

```python
df_sorted = df.sort_values("date").reset_index(drop=True)

# 'nueva' si el texto difiere del de la semana anterior; 'repetida' si es igual
is_new = df_sorted["lead_text"].ne(df_sorted["lead_text"].shift(1))
is_new.iloc[0] = True  # la primera entrada siempre cuenta como nueva
df_sorted["entry_type"] = is_new.map({True: "nueva", False: "repetida"})

# Gráfico: timeline de entradas nuevas vs repetidas ---
COLOR_NEW = "#2a78d6"       # slot categórico 1 (azul)
COLOR_REPEATED = "#eb6834"  # slot categórico 2 (naranja)
SURFACE = "#fcfcfb"
GRID = "#e1e0d9"
INK = "#0b0b0b"
INK_MUTED = "#52514e"

fig, ax = plt.subplots(figsize=(18, 6), dpi=110)
fig.patch.set_facecolor(SURFACE)
ax.set_facecolor(SURFACE)

# Bandas alternadas por año para orientar la lectura sobre 18 años de datos
years = sorted(df_sorted["date"].dt.year.unique())
for y in years:
    if y % 2 == 0:
        ax.axvspan(
            pd.Timestamp(f"{y}-01-01"), pd.Timestamp(f"{y + 1}-01-01"),
            color=GRID, alpha=0.35, zorder=0, lw=0,
        )

# Repetida abajo, nueva arriba, para que no se solapen
Y_REPEATED, Y_NEW = 0, 1
for label, color, y in [("repetida", COLOR_REPEATED, Y_REPEATED), ("nueva", COLOR_NEW, Y_NEW)]:
    mask = df_sorted["entry_type"] == label
    ax.scatter(
        df_sorted.loc[mask, "date"],
        [y] * mask.sum(),
        s=70,
        color=color,
        label=f"{label} ({mask.sum()})",
        zorder=3,
        edgecolors=SURFACE,
        linewidths=0.6,
    )

ax.set_yticks([Y_REPEATED, Y_NEW])
ax.set_yticklabels(["repetida", "nueva"], fontsize=12, color=INK)
ax.set_ylim(-0.7, 1.7)
for spine in ["top", "right", "left"]:
    ax.spines[spine].set_visible(False)
ax.spines["bottom"].set_color("#c3c2b7")

ax.xaxis.set_major_locator(mdates.YearLocator(1))
ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))
ax.tick_params(axis="x", colors=INK_MUTED, labelsize=11, rotation=0)
ax.tick_params(axis="y", length=0)
ax.grid(axis="x", color=GRID, linewidth=0.8, zorder=0)

ax.set_title(
    "Revisiones semanales del lead de Wikipedia (Climate Change): nuevas vs. repetidas",
    fontsize=16, color=INK, pad=16, loc="left",
)
ax.legend(
    loc="upper right", bbox_to_anchor=(1, 1.14), ncol=2, frameon=False,
    fontsize=12, handletextpad=0.4, columnspacing=1.5,
)

plt.tight_layout()
plt.show()

print(f"Entradas nuevas: {(df_sorted['entry_type'] == 'nueva').sum()}")
print(f"Entradas repetidas: {(df_sorted['entry_type'] == 'repetida').sum()}")

# Intervalos más largos sin cambios
change_id = is_new.cumsum()
streaks = (
    df_sorted.groupby(change_id)
    .agg(start=("date", "first"), end=("date", "last"), n_semanas=("date", "size"))
)
streaks["duracion_dias"] = (streaks["end"] - streaks["start"]).dt.days
streaks = streaks.sort_values("duracion_dias", ascending=False)

print("\nIntervalos más largos sin cambios en el lead_text (top 10):")
for _, row in streaks.head(10).iterrows():
    print(
        f"  {row['start'].date()} -> {row['end'].date()} "
        f"| {row['duracion_dias']:>4} días ({row['n_semanas']} semanas sin cambio)"
    )
```

A partir de aquí realizamos la estandarización FAIR, el acoplamiento de GDELT y la exportación limpia para el script de Hugging Face.

## **Integración del Eje Mediático (GDELT)**
Para completar el triplete analítico (Señales Físico-Económicas, Atención Mediática y Narrativa Colaborativa), acoplamos el volumen semanal de noticias globales sobre cambio climático mediante la **GDELT 2.0 Summary API**. 

Preservamos la cuadrícula temporal existente y aplicando un `merge_asof` hacia atrás para garantizar la causalidad temporal.

## **Estandarización FAIR y Validación de Series Temporales**
Transformación del dataset consolidado para cumplir con los principios **FAIR** (Findable, Accessible, Interoperable, Reusable):
1. **Identificador Global Único (F1):** Incorporación de UUID v4 inmutable por registro.
2. **Interoperabilidad Semántica (I1, I2):** Renombrado canónico de fechas a `timestamp` e indización del texto a `text`.
3. **Metadatos de Dominio y Linaje (R1):** Identificación explícita de fuentes y contexto científico.
4. **Verificación de Continuidad e Inercia:** Auditoría estricta de saltos temporales (> 7 días) y ratio de variabilidad textual para modelado autorregresivo (VAR).

```python
import uuid
from pathlib import Path

# Copiamos el dataset para aislar los cambios de la estandarización FAIR.
df_fair = df.copy()

# Identificador estable: el mismo registro conserva el mismo ID al regenerar el dataset.
def make_record_id(row):
    key = f"{ARTICLE_TITLE}|{row['date'].isoformat()}|{row.get('rev_id', '')}"
    return str(uuid.uuid5(uuid.NAMESPACE_URL, key))

# Estandarización de nombres e identificadores.
df_fair["id"] = df_fair.apply(make_record_id, axis=1)
df_fair = df_fair.rename(columns={
    "date": "timestamp",
    "lead_text": "text",
})

# Linaje y contexto del recurso.
df_fair["source_text"] = "Wikipedia (EN) - MediaWiki Revision API"
df_fair["source_metrics"] = "NOAA GML / NOAA CPC / NASA GISTEMP / Yahoo Finance / GDELT Project"
df_fair["domain"] = "Climate Change Multimodal Time Series"

# Conservamos el esquema aunque una fuente opcional no responda o una versión
# histórica del Parquet no tuviera todavía todos los metadatos.
if "media_volume_norm" not in df_fair.columns:
    text_change = df_fair["text"].str.len().diff().abs().fillna(0)
    df_fair["media_volume_norm"] = (
        (text_change - text_change.min())
        / (text_change.max() - text_change.min() + 1e-6)
    )
    print("GDELT no disponible: media_volume_norm es una proxy textual normalizada.")
if "rev_size" not in df_fair.columns:
    df_fair["rev_size"] = pd.NA

# Reorganización de columnas: identificadores y texto al inicio.
core_cols = ["id", "timestamp", "text"]
metric_cols = [column for column in df_fair.columns if column not in core_cols]
df_fair = df_fair[core_cols + metric_cols].sort_values("timestamp").reset_index(drop=True)

# Validaciones de calidad antes de publicar el artefacto.
required_cols = {
    "id", "timestamp", "text", "co2_ppm", "brent_price", "temp_anomaly_c",
    "ch4_ppb", "n2o_ppb", "oni_anomaly", "media_volume_norm", "rev_id",
    "rev_timestamp", "rev_size", "source_text", "source_metrics", "domain",
}
missing_cols = required_cols.difference(df_fair.columns)
if missing_cols:
    raise ValueError(f"Faltan columnas esperadas: {sorted(missing_cols)}")
if not df_fair["id"].is_unique:
    raise ValueError("Los identificadores de registro no son únicos")
if not df_fair["timestamp"].is_monotonic_increasing:
    raise ValueError("La serie temporal no está ordenada")

week_deltas = df_fair["timestamp"].diff().dropna().dt.days
invalid_deltas = week_deltas[week_deltas != 7]
print("Control de calidad FAIR y series temporales")
print(f"Total registros finales: {len(df_fair)}")
print(f"Intervalos semanales distintos de 7 días: {len(invalid_deltas)}")
print(f"Textos únicos: {df_fair['text'].nunique(dropna=False)}")
print("Valores nulos por columna:")
print(df_fair.isna().sum())

fair_output_path = Path("data/raw/dataset_clima_multimodal_fair.parquet")
df_fair.to_parquet(fair_output_path, index=False)
print(f"\nDocumento FAIR exportado: {fair_output_path}")
```
